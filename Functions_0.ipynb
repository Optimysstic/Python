```python
# positional arguments

def nameage(name, age):
  print("hello i am:", name)
  print("my age is:", age)
print("case-1")
nameage("dush", 27)
print("\ncase-2")
nameage(27, "dush")
```

```python
# *args

def greet(*args):
  for name in args:
    print(f"hello, {name}!")
greet("dush")
greet("sambireddy", "sarada")
```

```python
# **kwargs

def person(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}: {value}")
person(name = "dush", age = 27, city = "princeton")
```

```python
# Mixing Regular + *args + **kwargs

def profile(role, *skills, **details):
  print(f"Role: {role}")
  print("Skills:", skills)
  print("Details:", details)
profile(
    "data scientist",
    "python", "ml", "deep learning",
    location = "new york city",
    experience = 2.5
)
```

Recursive Function

```python
# factorial number

def factorial(n):
  if n == 0 or n == 1: # base case
     return 1
  else: # recursive case
    return n * factorial(n-1)
print(factorial(7))
```

```python
# Fibonacci Sequence

def Fibonacci(n):
  if n == 0:
    return 0
  elif n == 1:
    return 1
  else:
    return Fibonacci(n-1) + Fibonacci(n-2)
print(Fibonacci(7))
```

```python
# Recursive Function for a List Sum

def list_sum(list):
  if len(list) == 0:
    return 0
  else:
    return list[0] + list_sum(list[1:])
print(list_sum([1,2,3,4]))
```

```python
# Tail Recursion (Factorial)
def tail_factorial(n, accumulator=1):
  if n == 0:
    return accumulator
  else:
    return tail_factorial(n-1, accumulator*n)
print(tail_factorial(7))
```

```python
# turning recursion into iteration

def factorial_iterative(n):
  acc = 1
  while n > 0:
    acc *= n
    n -= 1
  return acc
print(factorial_iterative(5))
```

In Python, this conversion is always preferred for large inputs because:

Python has no tail call optimization.

Recursion depth is limited (~1000 calls by default).

Loops are usually faster and more memory-efficient.

```python
# Print dush n times without the loop.

def name(n):
  if n == 0:
    return
  print("dush")
  name(n-1)
name(5)
```

```python
# making the same above code in single line

def name(n):
  n and (print("dush") or name(n-1))
name(5)
```
