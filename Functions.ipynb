```python
# Local Scope (Inside Function)

def foo():
    xx = 10
    print(xx)

foo()
# print(xx)
```

```python
# Global Scope (Outside Function)

x = 10

def foo():
  x = x + 1
  print(x)

foo()
```

```python
# Modifying Global Without Permission

x = 11

def foo():
    print(x)
foo()
```

```python
# global Keyword (Use Sparingly)

x = 10

def foo():
    global x
    x = x + 1

foo()
print(x)
```

Scope Traps (VERY COMMON MISTAKES)

```python
# Trap 1: Modifying Outer Variable (Without nonlocal)

def outer():
    x = 10
    def inner():
        x = x + 1
    inner()
outer()
```

```python
# Correct Way (nonlocal)

def outer():
    x = 10
    def inner():
        nonlocal x # 10
        x = x + 1 # 10 + 1 = 11
    inner() # 11
    print(x) # 11

outer()
```

```python
# Default Argument Evaluation (CRITICAL)

def add_item(item, list=[]):
    list.append(item)
    return list
add_item(1)
add_item(2)
```

```python
# Correct Pattern

def add_item(item, lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst
#add_item(1)
add_item(2)
```
